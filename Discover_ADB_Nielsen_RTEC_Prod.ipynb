```python
import glob as _glob
import os
import pandas as pd
import subprocess
from pyspark.sql.functions import (
    regexp_extract, regexp_replace, col,
    sum as spark_sum, round as spark_round, expr,
    to_date, date_format
)

_input_base_dir = "/Workspace/Users/abhishek.sawlikar@genmills.com/Discover_vs_ADB_Check/Input File/Nielsen/21-07-2026/RTEC"

# Auto-detect Discover file for Local_Market
# Primary:  {Local_Market}_Discover*.xlsx  e.g. Ireland_Discover_060726.xlsx
# Fallback: {Local_Market}*.xlsx           e.g. Kuwait_RI_Extract (4).xlsx
_primary  = _glob.glob(f"{_input_base_dir}/{Local_Market}_[Dd]iscover*.xlsx")
_fallback = _glob.glob(f"{_input_base_dir}/{Local_Market}*.xlsx") if not _primary else []
_found    = _primary or _fallback
if not _found:
    raise FileNotFoundError(
        f"No Discover file found for market '{Local_Market}' in:\n  {_input_base_dir}\n"
        f"Expected pattern: {Local_Market}_Discover*.xlsx"
    )

_detected_file     = os.path.basename(_found[0])
input_file_root    = _input_base_dir
discover_file_config = {Local_Market: {'file_name': _detected_file, 'file_type': 'excel'}}
markets_to_run     = list(discover_file_config.keys())
print(f"input_file_root : {input_file_root}")
print(f"markets_to_run  : {markets_to_run}")

# Resolve market config. Validates that the user-supplied Local_Market exists in a pre-defined mapping dictionary
detected_market = None
for _k, _v in local_market_mapping.items():
    if Local_Market.strip().upper() in (_k.upper(), _v.upper()):
        detected_market = _k
        break
if detected_market is None:
    raise ValueError(f"Unknown market: '{Local_Market}'. Add it to local_market_mapping and market_config.")
if detected_market not in discover_file_config:
    raise ValueError(
        f"No input file configured for market '{detected_market}'. "
        f"Configured markets: {sorted(discover_file_config.keys())}"
    )

subprocess.check_call(['pip', 'install', '-q', 'openpyxl'])

_file_cfg = discover_file_config[detected_market]
file_path = os.path.join(input_file_root, _file_cfg['file_name'])
file_type = _file_cfg['file_type']
print(f"Loading Discover file for {detected_market}: {file_path}")

# Load file — auto-detect header row for Excel files with metadata preambles
if file_type == 'csv':
    pdf = pd.read_csv(file_path)
else:
    _probe_raw = pd.read_excel(file_path, header=None, nrows=20)
    _all_known = {'Periods', 'Markets'} | {
        v for cfg in market_config.values()
        for v in [cfg['manufacturer_col'], cfg['value_col'], cfg['volume_col']]
    }
    _header_row = next(
        (_i for _i, _row in _probe_raw.iterrows()
         if any(str(v).strip() in _all_known for v in _row.values if pd.notna(v))),
        0
    )
    pdf = pd.read_excel(file_path, header=_header_row)

# Coerce mixed-type object columns to string for Arrow compatibility. Pandas object columns can contain mixed types (strings + numbers). This forces all non-null values in object columns to strings.
for _c in pdf.columns:
    if pdf[_c].dtype == object:
        pdf[_c] = pdf[_c].where(pdf[_c].isna(), pdf[_c].astype(str))

# Extract period from value column and filter manufacturer is not null records
df_cleansed = (
    spark.createDataFrame(pdf)
    .withColumn("Local_Period", regexp_extract("Periods", r"(\d{2}/\d{2}/\d{2})", 1))
    .withColumn(value_col, regexp_replace(col(f"`{value_col}`"), r"[^0-9.\-]", ""))
    .filter(col(f"`{manufacturer_col}`").isNotNull())
)

# Re-validate column names (guards against stale state from a prior run). If the expected column names (from a prior cell's variables) don't match the actual DataFrame columns, the code searches market_config for a configuration whose columns do exist in the data and auto-corrects. 
_disc_cols = set(df_cleansed.columns)
if volume_col not in _disc_cols or manufacturer_col not in _disc_cols or value_col not in _disc_cols:
    for _mkt, _mcfg in market_config.items():
        if (_mcfg['manufacturer_col'] in _disc_cols and
                _mcfg['value_col'] in _disc_cols and
                _mcfg['volume_col'] in _disc_cols):
            manufacturer_col = _mcfg['manufacturer_col']
            value_col        = _mcfg['value_col']
            volume_col       = _mcfg['volume_col']
            print(f"Auto-corrected columns to '{_mkt}' config "
                  f"(mfr={manufacturer_col}, val={value_col}, vol={volume_col})")
            break

# Use _discover_markets from Cell 3 — already scoped to the relevant ADB market descriptions
_adb_market_descs = _discover_markets if '_discover_markets' in globals() and _discover_markets else []
if _adb_market_descs:
    print(f"Filtering Discover to ADB market descriptions: {_adb_market_descs}")
else:
    print("Warning: _discover_markets not available — discover_agg will include all Markets rows.")

# discover_agg: columns aligned to adb_local (Local_Manufacturer, Value_000, Volume_000)
# so a direct join on Local_Manufacturer is possible without any renaming downstream
discover_agg = (
    (df_cleansed.filter(col("Markets").isin(_adb_market_descs)) if _adb_market_descs else df_cleansed)
    .withColumnRenamed(manufacturer_col, "Local_Manufacturer")
    .withColumn("Local_Time_Period", date_format(to_date(col("Local_Period"), "dd/MM/yy"), "yyyy-MM-dd"))
    .withColumn("Value_000",  expr(f"try_cast(`{value_col}`  as double)"))
    .withColumn("Volume_000", expr(f"try_cast(`{volume_col}` as double)"))
    .filter(col("Local_Manufacturer").isNotNull())
    .groupBy("Local_Manufacturer", "Local_Time_Period")
    .agg(
        spark_round(spark_sum("Value_000"),  4).alias("Value_000"),
        spark_round(spark_sum("Volume_000"), 4).alias("Volume_000")
    )
)

print(f"Local_Market filter  : {local_market}")
print(f"Markets filtered     : {_adb_market_descs}")
print(f"Discover manufacturers loaded: {discover_agg.count()}")
display(discover_agg)
```

```python
from pyspark.sql.functions import col, sum as spark_sum, round as spark_round, lit, coalesce, expr, date_format, to_date

# ============================================================
# STEP 4: MANUFACTURER-LEVEL DISCOVER vs ADB COMPARISON
# Uses from Cell 3 : local_market, markets_in, min_period,
#                    local_category_col, _sub_cat_filter,
#                    value_col, volume_col
# Uses from Cell 7 : max_period
# Schema both sides: Local_Manufactuer | Value_000 | Volume_000
# ============================================================

# 1. Aggregate Discover by period
#    Reuse Value_000/Volume_000 names so disc.*/adb.* aliases
#    disambiguate cleanly at join time (consistent with Cell 7).
discover_manufacturer_agg = (
    df_cleansed
    .withColumnRenamed(manufacturer_col, 'Local_Manufacturer')
    .withColumn('Local_Time_Period', date_format(to_date(col('Local_Period'), 'dd/MM/yy'), 'yyyy-MM-dd'))
    .filter(col('Local_Time_Period').between(min_period, max_period))
    .withColumn('Value_000',  expr(f'try_cast(`{value_col}`  as double)'))
    .withColumn('Volume_000', expr(f'try_cast(`{volume_col}` as double)'))
    .groupBy('Local_Manufacturer', 'Local_Time_Period')
    .agg(
        spark_round(spark_sum('Value_000'),  4).alias('Value_000'),
        spark_round(spark_sum('Volume_000'), 4).alias('Volume_000'),
    )
)

# 2. ADB manufacturer-level query
#    - local_category_col and _sub_cat_filter come directly from
#      Filters_config (set in Cell 3) -- no stale filter variables.
#    - Period formatted as dd/MM/yy to match Discover's Local_Period,
#      so the join is a plain string equality (no date type mismatch).
adb_manufacturer = spark.sql(f"""
SELECT
    p.Local_Manufacturer,
    date_format(t.Local_Time_Period, 'yyyy-MM-dd') AS Local_Time_Period,
    SUM(f.Value_000)  AS Value_000,
    SUM(f.Volume_000) AS Volume_000
FROM glbl_cpw_prod.conformed.factretailsales AS f
JOIN glbl_cpw_prod.conformed.dimproduct_local p
    ON f.LocalProductKey = p.LocalProductKey
JOIN glbl_cpw_prod.conformed.dimmarket gm
    ON f.LocalMarketKey = gm.LocalMarketKey
JOIN glbl_cpw_prod.conformed.dimperiod_local t
    ON f.LocalPeriodKey = t.LocalPeriodKey
WHERE gm.Local_Market = '{local_market}'
  AND gm.Local_Market_Description IN ({', '.join(f"'{d}'" for d in _all_adb_descs)})
  AND t.Local_Time_Period >= '{min_period}'
  AND t.Local_Time_Period <= '{max_period}'
  AND gm.Global_Total_Mkt_Flag = 'Y'
  AND p.Local_Manufacturer IS NOT NULL
  AND p.Local_Category IN ('{local_category_col}')
  {_sub_cat_filter}
GROUP BY p.Local_Manufacturer, t.Local_Time_Period
ORDER BY p.Local_Manufacturer, t.Local_Time_Period
""")

# 3. Inner join on the shared Local_Manufacturer and Local_Time_period
comparison_manufacturer = (
    discover_manufacturer_agg.alias('disc')
    .join(adb_manufacturer.alias('adb'), ['Local_Manufacturer', 'Local_Time_Period'], 'inner')
)

# 4. Result layout: Local_Manufacturer | Discover | ADB | Diff
result_manufacturer = comparison_manufacturer.select(
    col('Local_Manufacturer'),
    # Discover
    spark_round(coalesce(col('disc.Value_000'),  lit(0)), 4).alias(f'Sum of {value_col}'),
    spark_round(coalesce(col('disc.Volume_000'), lit(0)), 4).alias(f'Sum of {volume_col}'),
    # ADB
    spark_round(coalesce(col('adb.Value_000'),   lit(0)), 4).alias('Value_000'),
    spark_round(coalesce(col('adb.Volume_000'),  lit(0)), 4).alias('Volume_000'),
    # Diff: Discover - ADB
    spark_round(
        coalesce(col('disc.Value_000'),  lit(0)) - coalesce(col('adb.Value_000'),  lit(0)), 4
    ).alias('Diff_Value_000'),
    spark_round(
        coalesce(col('disc.Volume_000'), lit(0)) - coalesce(col('adb.Volume_000'), lit(0)), 4
    ).alias('Diff_Volume_000'),
).groupBy('Local_Manufacturer').agg(
    spark_round(spark_sum(f'Sum of {value_col}'),  4).alias(f'Sum of {value_col}'),
    spark_round(spark_sum(f'Sum of {volume_col}'), 4).alias(f'Sum of {volume_col}'),
    spark_round(spark_sum('Value_000'),            4).alias('Value_000'),
    spark_round(spark_sum('Volume_000'),           4).alias('Volume_000'),
    spark_round(spark_sum('Diff_Value_000'),       4).alias('Diff_Value_000'),
    spark_round(spark_sum('Diff_Volume_000'),      4).alias('Diff_Volume_000'),
).orderBy('Local_Manufacturer')

_sc_info = f' | Sub-cat filter: {_sub_cat_filter}' if _sub_cat_filter else ''
print(f'Manufacturer-level comparison for market : {detected_market}')
print(f'Category filter                    : Local_Category = {local_category_col!r}{_sc_info}')
display(result_manufacturer)
```
